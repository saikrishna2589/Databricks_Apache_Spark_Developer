````python
schema = 'id LONG, source STRING , destination STRING , distance LONG'

flights_routes_data = ([
                        (101, 'Mumbai', 'Goa', 587), 
                        (102, 'Mumbai', 'Bangalore', 985), 
                        (102, 'Mumbai', 'Bangalore', 985), 
                        (103, 'Delhi', 'Chennai', 2208), 
                        (104, 'Delhi', 'Chennai', 2208), 
                        (105, 'Bangalore', 'Kolkata', 1868),
                        (105, 'Bangalore', 'Kolkata', 1865),
                        ])



flights_routes_df = spark.createDataFrame(flights_routes_data ,schema)
````

````python
flights_routes_df.display()
````

####2. Select records that satisfy the follwing criteria
1. Source is Mumbai and destination is Bangalore

Show the following approaches
1. Using a single filter
2. Using two filters
3. Using a col function expression
4. Using a dataframe qualifier expression

````python
from pyspark.sql.functions import col,lower
flights_routes_mum_ban_df = flights_routes_df.filter((col('source')=='Mumbai') & (lower(col('destination'))== 'bangalore'))
````

````python
flights_routes_mum_ban_df.display()
````

````python
# use expr

flights_routes_mum_ban_df_expr = flights_routes_df.where("source = 'Mumbai' AND lower(destination) = 'bangalore'")
````

````python
flights_routes_mum_ban_df_expr 
````

````python
from pyspark.sql.functions import expr
flights_routes_mum_ban_df_expr =flights_routes_df.where(expr("source = 'Mumbai' AND lower(destination) = 'bangalore'"))
````

````python
flights_routes_mum_ban_df_expr.display()
````
