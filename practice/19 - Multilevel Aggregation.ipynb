```python
#step 1 : create dataframe and view it.
from pyspark.sql import Row

# Realistic e-commerce sales data
sales_data = [
    Row(region="North", category="Laptops", sales=15000),
    Row(region="North", category="Phones", sales=8000),
    Row(region="North", category="Tablets", sales=4000),
    Row(region="South", category="Laptops", sales=12000),
    Row(region="South", category="Phones", sales=9000),
    Row(region="South", category="Tablets", sales=3000),
    Row(region="East", category="Laptops", sales=18000),
    Row(region="East", category="Phones", sales=11000),
    Row(region="West", category="Laptops", sales=10000),
    Row(region="West", category="Tablets", sales=5000)
]
```

```python
sales_df = spark.createDataFrame(sales_data)
```

```python
display(sales_df)
```

```python
from pyspark.sql.functions import sum,col 
sales_rollup = sales_df.rollup("region","category").agg(sum(col('sales')).alias('total_sales'))
```

```python
sales_rollup.orderBy('region','category').display()
```

```python
#cube : step 3

sales_cube = sales_df.cube("region", "category").agg(sum(col("sales")).alias("total_sales"))
```

```python
sales_cube.orderBy("region","category").display()
```

```python
sales_gsets = (sales_df.groupingSets
               #outer list
               ([
                   [col("region"), 
                 col("category")],
                [col("region")], 
                [col("category")]
               ],
                #columns to display
                col("region"),col("category"))
               .agg(sum(col("sales")).alias("total_sales")))
```

```python
sales_gsets.orderBy('region',
                'category').display()
```
