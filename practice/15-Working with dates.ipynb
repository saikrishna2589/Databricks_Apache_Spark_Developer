###### Working with Dates

```python
data_list = [(2022, 5, 18) , (9999, 12, 31), (-9999, 1, 1), (10000, 1, 1), 
             (-10000, 1, 1), (193, 5, 25),   (99, 5, 25),   (1000, 2, 29)]
```

```python
dates_learn_df = spark.createDataFrame(data_list).toDF('Y','M','D')
```

```python
dates_learn_df.display()
```

```python
from pyspark.sql.functions import concat_ws
learndate_df = dates_learn_df.withColumn('date_str',concat_ws("-",*dates_learn_df.columns))
```

```python
learndate_df.display()
```

```python
from pyspark.sql.functions import to_date, try_to_date,col, expr
```

```python
learndate_df_str_to_date_conversion=learndate_df.withColumn(
    "date_str_try_date", 
    try_to_date(col("date_str"),"y-M-d"
))
```

```python
from pyspark.sql.functions import date_add,date_sub,current_date
```

```python
learndate_df_str_to_date_conversion.display()
```

```python
learndate_df_date_transformations =learndate_df_str_to_date_conversion.withColumns(
    {
   "add_10_days_from_today": 
       date_add(current_date() , 10),
    "add_7_days" : date_add(
        "date_str_try_date", 7), 
    "subtract 10 days from current" : date_add(current_date(), -10),
    "subtract 10 days" : date_add("date_str_try_date",-10)

    }
)
```

```python
learndate_df_date_transformations.display()
```

```python
from pyspark.sql.functions import date_diff, add_months,date_part,lit,date_format
```

```python
learndate_df_date_transformations=learndate_df_date_transformations.withColumns({

    "date_diff" : date_diff(current_date(), "date_str_try_date"),
    "yearpart":date_part(lit("YEAR") ,"date_str_try_date"),
     "monthpart":date_part(lit("month") ,"date_str_try_date"),
     "date_format_string" : date_format("date_str_try_date", "yy-MMM-d")
   
})
```

```python
learndate_df_date_transformations.display()
```
