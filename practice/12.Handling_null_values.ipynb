####Handling Null values
1. Equality check
2. Null in expressions
3. [Conditional functions for Null](https://spark.apache.org/docs/latest/api/python/reference/pyspark.sql/functions.html#conditional-functions)
4. Null in aggregations

####1. Create a data frame to demo some scenarios

```python
person_list = [(100, "Prashant", 30),
             (101, "David", None),
             (102, "Sushant",None),
             (103, "Abdul", 45),
             (104, "Shruti", 28)]

df = spark.createDataFrame(data = person_list).toDF('id','name','age')

df.display()
             
```

####2. How null equality is executed.

2.1 Find all records where age is 28.\
2.1 Find all records where age is not given or unknown.

```python
df_age_filter = df.where("age==28")
```

```python
df_age_filter = df.filter("age==28")
```

```python
from pyspark.sql.functions import col
df_age_filter = df.filter(col("age")==28)
df_age_filter.display()
```

```python
df_nulls_check = df.filter(col("age").isNull())
```

```python
df_nulls_check.display()
```

```python
df_nulls_condition_column = df.withColumn("isNull?" ,col("age").isNull())
```

```python
df_nulls_condition_column.display()
```

```python
from pyspark.sql.functions import expr
#df.withColumn("isNull?" ,expr("age IS NULL")).display()

df.filter(expr('age IS NULL')).display()  #sql way
df.filter(col("age").isNull()).display() # pyspark way
```

2.3 Use case for null equality

Select only those persons having a valid age information

```python
df.filter(col("age").isNotNull()).display() # pyspark way
```

####3. How operators work on null values



```python
from pyspark.sql.functions import expr
df.filter(col("age")>25).where(expr("name == 'Prashant'")).display()
```

3.3 How mathametical operators work on null.\
Calculate experience for every employee using the following formula.\
experience = age - 23


```python
df.withColumn('experience', expr("nvl(age,23)-23") ).display()
```

```python
#aggreagte

df.selectExpr("avg(age)").display()
```

```python
from pyspark.sql.functions import avg
df.filter(col("age").isNotNull()).select(avg(col("age"))).display()

```
