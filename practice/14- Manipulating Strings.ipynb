####3. Requirement
You are given a dataframe below
```
  df_address= spark.createDataFrame([("Benga`li Market", "110001"),("Adu~godi", "560030")]) \
          .toDF("address", "pin")
```
Write code to fix the data problem in the address field

````python
df_address= spark.createDataFrame([("Benga`li Market", "110001"),("Adu~godi", "560030")]) \
          .toDF("address", "pin")
````

````python
df_address.display()
````

````python
from pyspark.sql.functions import regexp_replace,col

df_address_trimmed = (df_address.select(
    regexp_replace(
        col('address'),
           "[~`]+",
            ""),
    col("pin")
)
)
````

````python
df_address_trimmed.display()
````
