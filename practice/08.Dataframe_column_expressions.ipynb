Requirement
Read data from flight_time and transform as below:


- Rename fl_date to dep_date
- Compute arr_date
- Following fields to represent full timestamp
- crs_dep_time
- dep_time
- crs_arr_time
- arr_time


```python
%sql
select * from dev.spark_db.flight_time
```

```python
flight_schedule_df =spark.read.table("dev.spark_db.flight_time")
```

```python
from pyspark.sql.functions import col
flight_schedule_df = flight_schedule_df.withColumnRenamed('Fl_Date', 'departure_date').withColumn("arrival_date" ,col('departure_date')+col('DEP_TIME_TIME_IN_HR_MIN')+col(
'WHEELS_ON_IN_HR_MIN'))
```

```python
flight_schedule_df.display()
```

```python
from pyspark.sql import functions as F
flight_schedule_df = flight_schedule_df.withColumn("arrival_date", F.to_date(F.col("arrival_date")))\
                                .withColumn("arrival_date", F.when(F.col("arrival_date").isNull(), F.col("departure_date")).otherwise(F.col("arrival_date")))
```

```python

flight_schedule_df_updated =flight_schedule_df.selectExpr('CAST(departure_date AS TIMESTAMP)+ CRS_DEP_TIME_IN_HR_MIN AS depature_datetime', '*')
```
