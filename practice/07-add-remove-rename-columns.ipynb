####1. Create a dataframe as below
```
  +---+--------+---+------+
  | id|    name|age|salary|
  +---+--------+---+------+
  |100|Prashant| 45| 45000|
  |101|   Tarun| 36| 33000|
  |102|   David| 48| 28000|
  +---+--------+---+------+
```

````python
schema_df = 'id long, name string, age long, salary long'
````

````python
#create dataframe is list of tuples
data = ([(100,'Prashanth',45,45000),
        (101,'Tarun',45,33000),
        (102,'David',48,28000)])


schema_df = 'id integer,name string, age integer,salary integer '


df = spark.createDataFrame(data, schema_df)
````

````python
df.display()
````

####2. Add following columns to your Dataframe
1. increment: 10% of the salary up to 3000 maximum increment
2. revised_salary: salary + increment

````python
from pyspark.sql.functions import when,col
df_increment = df.withColumn('increment', when(col('salary')*.1>=3000 , 3000)\
    .otherwise(col('salary')*.1))
````

````python
df_increment.display()
````

````python
revised_salary = df_increment.withColumn('revised_salary',col('salary') + col('increment'))
````

````python
revised_salary.display()
````

````python
revised_salary_adjustment = revised_salary.withColumn('salary', col('revised_salary'))
````

````python
revised_salary_adjustment.display()
````

4. Add a Batch number(uuid) column to the df.

````python
import pyspark.sql.functions as F #this is pyspark fucntion
import uuid #this uuid module is a python module , not pyspark

batch_id_python_method = uuid.uuid4()

revised_salary_adjustment=  (
    revised_salary_adjustment
     .withColumns({'batch_number_random_pyspark': F.uuid(),
                   'batch_number_python_method': F.lit(str(batch_id_python_method))})
                                )
````

````python
revised_salary_adjustment.display()
````
